```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df_train = pd.read_csv ("train.csv")
df_test = pd.read_csv ("test.csv") 
```

```python
df_train.head(3)
```

```python
df_test.head(3)
```

```python
# Describir brevemente la cantidad de valores faltantes para cada una de las variables de la base de datos.
print ("--- valores faltantes en train ----")
print(df_train.isnull().sum()[df_train.isnull().sum()>0])
print ("--- valores faltantes en test ---")
print(df_test.isnull().sum()[df_test.isnull().sum()>0])
```

```python
#Completar aquellas variables que se encuentran faltantes para las bases de train (“Age”,”fare”, “Cabin”) y test ( “Age”, “Cabin”, “embarked”).

mean_age_train = df_train['Age'].mean()
print (f" -- rellenado 'Age' con el el promedio {mean_age_train:.2f}")
mean_age_test =df_test['Age'].mean()
print (f"-- rellenado 'Age' con el promedio {mean_age_test:.2f}")
df_train['Age'] = df_train['Age'].fillna(mean_age_train)
df_test['Age'] = df_test['Age'].fillna(mean_age_test)
```

```python
mean_fare_train = df_train['Fare'].mean()
print (f" -- rellenado 'Fare' con el el promedio {mean_fare_train:.2f}")
mean_fare_test =df_test['Fare'].mean()
print (f"-- rellenado 'Fare' con el promedio {mean_fare_test:.2f}")
df_train['Fare'] = df_train['Fare'].fillna(mean_fare_train)
df_test['Fare'] = df_test['Fare'].fillna(mean_fare_test)
```

```python
mode_embarked_train = df_train["Embarked"].mode()[0]
print (f" -- rellenado de 'Embarked' con la moda: {mode_embarked_train}")
df_train['Embarked'] = df_train['Embarked'].fillna(mode_embarked_train)
```

```python
total_files_train = len(df_train)
nulos_cabin_train =df_train ['Cabin'].isnull().sum()
porcentaje_train = (nulos_cabin_train / total_files_train) * 100
print(f"porcentaje de 'Cabin' faltante en train.csv: {porcentaje_train:.2f}%")

total_files_test = len(df_test)
nulos_cabin_test =df_test ['Cabin'].isnull().sum()
porcentaje_test = (nulos_cabin_test / total_files_test) * 100
print(f"porcentaje de 'Cabin' faltante en test.csv: {porcentaje_test:.2f}%")
```

```python
print("-- Eliminado de columna 'Cabin' por porcentaje alto de datos faltantes")
df_train = df_train.drop('Cabin', axis=1)
df_test = df_test.drop('Cabin', axis=1)
```

```python
df_train['Sex'] =df_train['Sex'].map ({'male':0, 'female':1})
df_test['Sex'] =df_test['Sex'].map ({'male':0, 'female':1})
```

```python
df_train = pd.get_dummies(df_train, columns=['Embarked'])
df_test = pd.get_dummies(df_test, columns=['Embarked'])
```

```python
#Ajustar el primer modelo de regresión logística
print ("Ajuste de modelo de regresion")
x = df_train_processed.drop("Survived", axis=1)
y= df_train_processed['Survived']
```

```python
x_train,X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=28)
print (f'Datos de entrenamiento divididos: {len(x_train)} para entrenar, {len(X_test)} para probar.')
```

```python
model = LogisticRegression(max_iter=2000)
```

```python
model.fit(x_train, y_train)
print ('Modelo entrenado exitosamente')
```
